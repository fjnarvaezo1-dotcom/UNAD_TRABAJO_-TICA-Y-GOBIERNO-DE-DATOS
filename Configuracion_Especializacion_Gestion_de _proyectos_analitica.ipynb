## Este codigo muestra las anomalias temporales con los leads

```python
import pandas as pd

# df es el dataframe original de leads del notebook
df["contact_date"] = pd.to_datetime(
df["contact_date"],
errors="coerce"
)

# Localizar el dataframe de campañas por su estructura,
# sin inventar el nombre utilizado en el notebook
columnas_campanas = {
"campaign_id",
"campaign_name",
"start_date",
"end_date",
"impressions",
"clicks",
"cost_usd",
"region"
}

dataframes_campanas = [
objeto
for objeto in globals().values()
if isinstance(objeto, pd.DataFrame)
and objeto is not df
and columnas_campanas.issubset(objeto.columns)
]

if not dataframes_campanas:
    raise ValueError(
    "No se encontró el dataframe original de campañas."
    )

# Referencia al dataframe original de campañas
campanas = dataframes_campanas[0].copy()

campanas["start_date"] = pd.to_datetime(
campanas["start_date"],
errors="coerce"
)

campanas["end_date"] = pd.to_datetime(
campanas["end_date"],
errors="coerce"
)

# Cruce entre los leads y las campañas
df_validacion = df.merge(
campanas[
[
"campaign_id",
"campaign_name",
"start_date",
"end_date"
]
],
on="campaign_id",
how="left",
validate="many_to_one"
)

# Un contacto es anómalo si ocurre antes del inicio
# o después de la finalización de la campaña
df_validacion["anomalia_temporal"] = (
(df_validacion["contact_date"] < df_validacion["start_date"])
|
(df_validacion["contact_date"] > df_validacion["end_date"])
)

# Diferencia de días frente al periodo válido
df_validacion["dias_antes_del_inicio"] = (
df_validacion["start_date"]
- df_validacion["contact_date"]
).dt.days

df_validacion["dias_despues_del_fin"] = (
df_validacion["contact_date"]
- df_validacion["end_date"]
).dt.days

# Registros con anomalías
anomalias_temporales = df_validacion.loc[
df_validacion["anomalia_temporal"],
[
"lead_id",
"campaign_id",
"campaign_name",
"contact_date",
"start_date",
"end_date",
"dias_antes_del_inicio",
"dias_despues_del_fin"
]
].sort_values(
["contact_date", "campaign_id"]
)

anomalias_temporales
```

```python
import pandas as pd

# 1. Métricas por fila (nivel campaña/lead directo)
derged_2['roi_marketing_pct'] = ((derged_2['conversion_value'] - derged_2['cost_usd']) / derged_2['cost_usd']) * 100
derged_2['conversion_rate_ads_pct'] = (derged_2['conversions'] / derged_2['clicks']) * 100

# 2. Métricas globales de Marketing (deduplicando campañas para no inflar costos)
df_campanas = derged_2.drop_duplicates(subset=['campaign_id'])
roi_ads_global = ((df_campanas['conversion_value'].sum() - df_campanas['cost_usd'].sum()) / df_campanas['cost_usd'].sum()) * 100
cr_ads_global = (df_campanas['conversions'].sum() / df_campanas['clicks'].sum()) * 100

# 3. Métricas del Embudo de Ventas (CRM)
cr_cierre_leads = derged_2['closed_deal'].mean() * 100
ingresos_cerrados = derged_2.loc[df['closed_deal'] == True, 'deal_value'].sum()
roi_ventas_real = ((ingresos_cerrados - df_campanas['cost_usd'].sum()) / df_campanas['cost_usd'].sum()) * 100

print(f"ROI Marketing (Deduplicado): {roi_ads_global:.2f}%")
print(f"Conversion Rate Ads (Deduplicado): {cr_ads_global:.2f}%")
print(f"Conversion Rate Cierre de Leads: {cr_cierre_leads:.2f}%")
print(f"ROI Ventas Cerradas: {roi_ventas_real:.2f}%")
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression

# ==========================================
# 1. PREPARACIÓN Y DEDUPLICACIÓN DE CAMPAÑAS
# ==========================================
# Asumiendo que tu DataFrame original se llama 'df'
camp_df = df.drop_duplicates(subset=['campaign_id']).copy()

# Creación de métricas de rendimiento por campaña
camp_df['ROI'] = ((camp_df['conversion_value'] - camp_df['cost_usd']) / camp_df['cost_usd']) * 100
camp_df['CR_Ads'] = (camp_df['conversions'] / camp_df['clicks']) * 100
camp_df['CPC'] = camp_df['cost_usd'] / camp_df['clicks']
camp_df['CPA'] = camp_df['cost_usd'] / camp_df['conversions']

num_cols = [
    'cost_usd', 'impressions', 'clicks', 'conversions',
    'conversion_value', 'leads', 'ROI', 'CR_Ads', 'CPC', 'CPA'
]

# ==========================================
# 2. ESTANDARIZACIÓN (Z-SCORE) Y MODELOS
# ==========================================
scaler = StandardScaler()
camp_std = pd.DataFrame(
    scaler.fit_transform(camp_df[num_cols]),
    columns=num_cols,
    index=camp_df.index
)

# Modelo 1: Regresión Lineal para proyectar el ROI en función de CR_Ads y CPC
X_roi = camp_df[['CR_Ads', 'CPC']]
y_roi = camp_df['ROI']
reg_roi = LinearRegression().fit(X_roi, y_roi)

# Simulación de escenarios (+5% relativo y +5% absoluto en Conversion Rate)
cr_actual = camp_df['CR_Ads'].mean()
cpc_actual = camp_df['CPC'].mean()

escenarios = pd.DataFrame({
    'CR_Ads': [cr_actual, cr_actual * 1.05, cr_actual + 5.0],
    'CPC': [cpc_actual, cpc_actual, cpc_actual]
}, index=['Actual', '+5% Relativo', '+5% Absoluto'])

escenarios['ROI_Proyectado'] = reg_roi.predict(escenarios[['CR_Ads', 'CPC']])

# Modelo 2: Regresión sobre datos estandarizados para hallar qué impulsa el CR_Ads
features_control = ['CPC', 'clicks', 'impressions', 'cost_usd']
reg_cr_std = LinearRegression().fit(camp_std[features_control], camp_std['CR_Ads'])
pesos_std = pd.Series(reg_cr_std.coef_, index=features_control).sort_values()

# ==========================================
# 3. CONFIGURACIÓN DE GRÁFICOS (3 PANELES)
# ==========================================
sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 3, figsize=(20, 6))

# --- PANEL 1: Mapa de Calor de Correlación de Pearson ---
corr_matrix = camp_std.corr(method='pearson')
sns.heatmap(
    corr_matrix, annot=True, fmt=".2f", cmap="coolwarm",
    vmin=-1, vmax=1, linewidths=0.5, cbar_kws={"shrink": 0.8}, ax=axes[0]
)
axes[0].set_title("1. Matriz de Correlación de Pearson\n(Datos Estandarizados)", fontsize=12, fontweight='bold')

# --- PANEL 2: Línea de Regresión y Proyección de ROI (+5% CR) ---
cr_grid = np.linspace(camp_df['CR_Ads'].min(), 15, 100)
grid_df = pd.DataFrame({'CR_Ads': cr_grid, 'CPC': cpc_actual})
roi_linea = reg_roi.predict(grid_df)

axes[1].scatter(camp_df['CR_Ads'], camp_df['ROI'], alpha=0.45, color='#64748b', label='Campañas Únicas (46)')
axes[1].plot(cr_grid, roi_linea, color='#2563eb', linewidth=2.5, label='Recta de Regresión (CPC cte.)')

# Puntos de simulación
colores_sim = ['#dc2626', '#f59e0b', '#16a34a']
for (nombre, fila), color in zip(escenarios.iterrows(), colores_sim):
    axes[1].scatter(
        fila['CR_Ads'], fila['ROI_Proyectado'], color=color, s=110, zorder=5,
        label=f"{nombre}: CR {fila['CR_Ads']:.2f}% → ROI {fila['ROI_Proyectado']:.1f}%"
    )

axes[1].set_title("2. Regresión Lineal: Proyección de ROI\nante aumento del +5% en Conversion Rate", fontsize=12, fontweight='bold')
axes[1].set_xlabel("Conversion Rate Ads (%)")
axes[1].set_ylabel("ROI de Marketing (%)")
axes[1].legend(loc='upper left', fontsize=9, frameon=True)

# --- PANEL 3: Importancia de Variables Estandarizadas (Pesos Beta) ---
colores_barras = ['#ef4444' if v < 0 else '#10b981' for v in pesos_std.values]
pesos_std.plot(kind='barh', ax=axes[2], color=colores_barras, edgecolor='black')
axes[2].axvline(0, color='black', linestyle='--', linewidth=1)
axes[2].set_title("3. Pesos Estandarizados (Beta)\n¿Qué variables aumentan el Conversion Rate?", fontsize=12, fontweight='bold')
axes[2].set_xlabel("Impacto en Desviaciones Estándar sobre CR_Ads")

for i, v in enumerate(pesos_std.values):
    axes[2].text(v + (0.03 if v >= 0 else -0.15), i, f"{v:.3f}", va='center', fontweight='bold')

plt.tight_layout()
plt.savefig("modelo_regresion_correlacion.png", dpi=150)
plt.show()
```
