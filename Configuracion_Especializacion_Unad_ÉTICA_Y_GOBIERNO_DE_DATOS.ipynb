### Consolidacion de datos y analisis , se realizo una unificacion a continuacion se confirman los datos y se concatenan despues de ajustar el dataframe de los Datos.

```python
derged = pd.merge(df, df_2, on='campaign_id', how='inner')
derged_2 = pd.DataFrame(derged)
derged_2.head(100)
```

### Consolidacion y limpieza de Datos los cuales se consolidaron para extraer las metricas de la informacion con lo cual se implemeto una IA, para realizar el analisis de los datos, en el siguiente dataframe se confirman los datos los cuales estan ya correlacionados y limpios para el proceso de analisis.

```python
data = derged_2[["lead_id","campaign_id","closed_deal","responded","scheduled_demo","impressions","clicks","cost_usd","deal_value","region"]]
data = pd.DataFrame(data)
data.head(500)
```

### Costo de las campañas y los leads por region, resumen ejecutivo despues de limpiar los datos.

```python
information_coust = data.groupby(["region"])["cost_usd"].sum()
information_coust = pd.DataFrame(information_coust)
information_coust
```
